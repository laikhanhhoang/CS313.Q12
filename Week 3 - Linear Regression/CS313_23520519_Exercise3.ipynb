```python
from sklearn.datasets import fetch_openml
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

# I. Chuẩn bị dữ liệu
Dataset Boston Housing có sẵn để dự đoán giá nhà (MEDV) ở khu vực Boston

```python
boston = fetch_openml(name='boston', version=1, as_frame=True)
df = boston.frame
print(df.shape)
df.head()
```

```python
X, y = df.drop(columns="MEDV"), df["MEDV"]
scaler = StandardScaler()
X = scaler.fit_transform(X)
print(X.shape, y.shape)
```

# II. Linear Regression với Sk-learn

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=False)
model.fit(X, y)
print("Hệ số (beta):", model.coef_)
```

# III. Normal Equation

```python
X_np, y_np = X.astype(float), y.astype(float).to_numpy()
print(X_np.shape, y_np.shape)
```

```python
beta_OLS = X_np.T @ X_np
beta_OLS = np.linalg.inv(beta_OLS)
beta_OLS = beta_OLS @ X_np.T
beta_OLS = beta_OLS @ y_np
```

```python
print("Hệ số (beta):", beta_OLS)
```

# IV. Gradient Descent

## Xác định khoảng learning rate đảm bảo hội tụ dựa trên hằng số Lipschitz

```python
X_np = X
n = X_np.shape[0]

# Lipschitz constant L = max eigenvalue của X^TX / n
L = np.linalg.eigvals(X_np.T @ X_np / n).max()

# Chọn learning rate < 2 / L
lr_max = 2 / L
print("lr_max =", lr_max)
```

```python
iterations = 300000
lr = 0.3264
n = X.shape[0]
p = X.shape[1]
beta_GD = np.random.randn(p)
```

## Thực hiện Gradient Descent

```python
for i in range(iterations):
    grad = X_np.T @ (X_np @ beta_GD - y_np) / n
    beta_GD = beta_GD - lr * grad
```

```python
print("Hệ số (beta):", beta_GD)
```
